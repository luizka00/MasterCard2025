```python
full_df['year_TIMESTAMP']         = full_df['timestamp'].dt.year
full_df['month_TIMESTAMP']        = full_df['timestamp'].dt.month
full_df['day_TIMESTAMP']          = full_df['timestamp'].dt.day
full_df['weekday_TIMESTAMP']      = full_df['timestamp'].dt.weekday
full_df['hour_TIMESTAMP']         = full_df['timestamp'].dt.hour
full_df['minute_TIMESTAMP']       = full_df['timestamp'].dt.minute
```

```python
full_df['signup_date_USER'] = pd.to_datetime(
    full_df['signup_date_USER'],
    format='%Y-%m-%d',    
    errors='coerce'       
)
```

```python
full_df['time_since_signup_TIMESTAMP_SIGNUP'] = full_df['timestamp'] - full_df['signup_date_USER']
```
